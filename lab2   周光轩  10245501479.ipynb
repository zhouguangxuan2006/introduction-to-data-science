```python
# 房屋价格数据集预处理实验

## 1. 导入必要的库
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.impute import KNNImputer
from sklearn.preprocessing import StandardScaler, KBinsDiscretizer
from sklearn.ensemble import RandomForestRegressor
import warnings
warnings.filterwarnings('ignore')

# 设置可视化风格
sns.set_style('whitegrid')
sns.set(font_scale=1.2)
## 2. 加载数据集# 读取数据
df = pd.read_csv('data/train.csv')

# 查看数据基本信息
print(f"数据集形状: {df.shape}")
print("\n数据集前5行:")
display(df.head())

# 查看数据类型和缺失值情况
print("\n数据集信息:")
df.info()

# 查看数值型特征的统计描述
print("\n数值型特征统计描述:")
display(df.describe())
## 3. 缺失值检测与处理# 检测缺失值
missing_values = df.isnull().sum()
missing_percentage = (missing_values / len(df)) * 100
missing_df = pd.DataFrame({
    '缺失值数量': missing_values,
    '缺失值比例(%)': missing_percentage
})
missing_df = missing_df[missing_df['缺失值数量'] > 0].sort_values('缺失值比例(%)', ascending=False)

print("缺失值统计:")
display(missing_df)

# 可视化缺失值
plt.figure(figsize=(15, 8))
sns.heatmap(df.isnull(), yticklabels=False, cbar=False, cmap='viridis')
plt.title('缺失值分布热力图')
plt.tight_layout()
plt.show()

# 分离数值型和类别型特征
numeric_features = df.select_dtypes(include=['int64', 'float64']).columns
categorical_features = df.select_dtypes(include=['object']).columns

# 处理类别型特征缺失值 - 用众数填充
for col in categorical_features:
    if df[col].isnull().sum() > 0:
        df[col].fillna(df[col].mode()[0], inplace=True)

# 处理数值型特征缺失值 - 使用KNN填充
numeric_df = df[numeric_features]
imputer = KNNImputer(n_neighbors=5)
imputed_numeric = imputer.fit_transform(numeric_df)

# 将填充后的数据放回到原始DataFrame
df[numeric_features] = imputed_numeric

# 验证是否还有缺失值
print(f"\n处理后缺失值总数: {df.isnull().sum().sum()}")
## 4. 异常值检测与处理# 定义异常值检测函数
def detect_outliers(df, n, features):
    """
    检测并返回异常值索引
    df: 数据集
    n: 标准差倍数阈值
    features: 要检测异常值的特征列表
    """
    outlier_indices = []
    
    for col in features:
        # 计算均值和标准差
        mean = df[col].mean()
        std = df[col].std()
        
        # 计算异常值边界
        lower_bound = mean - n * std
        upper_bound = mean + n * std
        
        # 找出异常值索引
        outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)].index
        outlier_indices.extend(outliers)
    
    # 计算每个索引出现的次数
    outlier_indices = pd.Series(outlier_indices).value_counts()
    # 选出出现次数超过1次的索引（在多个特征中都是异常值）
    outlier_indices = outlier_indices[outlier_indices > 1].index
    
    return outlier_indices

# 选择数值型特征检测异常值
outliers = detect_outliers(df, 3, numeric_features)
print(f"检测到的异常值数量: {len(outliers)}")

# 可视化异常值 - 以GrLivArea和SalePrice为例
plt.figure(figsize=(12, 6))
plt.scatter(df['GrLivArea'], df['SalePrice'], color='blue', label='正常数据')
plt.scatter(df.loc[outliers, 'GrLivArea'], df.loc[outliers, 'SalePrice'], color='red', label='异常值')
plt.xlabel('地上居住面积')
plt.ylabel('销售价格')
plt.title('GrLivArea与SalePrice的散点图（异常值标记）')
plt.legend()
plt.tight_layout()
plt.show()

# 处理异常值 - 用中位数填充
for col in numeric_features:
    df.loc[outliers, col] = df[col].median()

print("异常值处理完成")
## 5. 特征间相关性分析# 计算特征间的相关性
correlation = df[numeric_features].corr()

# 可视化相关性矩阵
plt.figure(figsize=(18, 15))
mask = np.triu(np.ones_like(correlation, dtype=bool))
sns.heatmap(correlation, mask=mask, annot=False, cmap='coolwarm', fmt=".2f", 
            square=True, linewidths=.5, cbar_kws={"shrink": .8})
plt.title('特征相关性热力图')
plt.tight_layout()
plt.show()

# 查看与SalePrice相关性最高的前10个特征
price_correlation = correlation['SalePrice'].sort_values(ascending=False)
print("与房价相关性最高的前10个特征:")
print(price_correlation[1:11])  # 排除自身

# 可视化相关性最高的几个特征与房价的关系
top_corr_features = price_correlation[1:4].index.tolist()  # 取前三名

plt.figure(figsize=(18, 5))
for i, feature in enumerate(top_corr_features, 1):
    plt.subplot(1, 3, i)
    sns.scatterplot(x=df[feature], y=df['SalePrice'])
    plt.title(f'{feature} 与 SalePrice 的关系 (相关系数: {price_correlation[feature]:.3f})')

plt.tight_layout()
plt.show()
## 6. 房价(SalePrice)标准化# 保存标准化前的房价用于对比
price_before = df['SalePrice'].copy()

# 标准化房价
scaler = StandardScaler()
df['SalePrice_scaled'] = scaler.fit_transform(df[['SalePrice']])

# 对比标准化前后的房价分布
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
sns.histplot(price_before, kde=True, color='blue')
plt.title('标准化前的房价分布')
plt.xlabel('房价')

plt.subplot(1, 2, 2)
sns.histplot(df['SalePrice_scaled'], kde=True, color='red')
plt.title('标准化后的房价分布')
plt.xlabel('标准化房价')

plt.tight_layout()
plt.show()

# 输出标准化前后的统计信息
print("标准化前的房价统计:")
print(price_before.describe())

print("\n标准化后的房价统计:")
print(df['SalePrice_scaled'].describe())
## 7. 房价离散化# 对房价进行离散化
discretizer = KBinsDiscretizer(n_bins=5, encode='ordinal', strategy='quantile')
df['SalePrice_binned'] = discretizer.fit_transform(df[['SalePrice']])

# 将离散化结果转换为类别
df['SalePrice_category'] = pd.cut(
    df['SalePrice'], 
    bins=discretizer.bin_edges_[0],
    labels=['极低', '低', '中', '高', '极高']
)

# 可视化离散化结果
plt.figure(figsize=(12, 6))

plt.subplot(1, 2, 1)
sns.countplot(x='SalePrice_binned', data=df)
plt.title('房价离散化结果（数值）')
plt.xlabel('离散化房价')

plt.subplot(1, 2, 2)
sns.countplot(x='SalePrice_category', data=df, order=['极低', '低', '中', '高', '极高'])
plt.title('房价离散化结果（类别）')
plt.xlabel('房价类别')

plt.tight_layout()
plt.show()

# 显示每个类别的范围
bin_edges = discretizer.bin_edges_[0]
print("房价离散化区间:")
for i in range(len(bin_edges) - 1):
    print(f"类别 {i} ({['极低', '低', '中', '高', '极高'][i]}): {bin_edges[i]:.2f} - {bin_edges[i+1]:.2f}")
## 8. 筛选房价相关性Top3特征并解释# 显示与房价相关性最高的三个特征
top3_features = price_correlation[1:4].index.tolist()
print(f"与房价相关性最高的三个特征: {top3_features}")
print("\n它们的相关系数:")
for feature in top3_features:
    print(f"{feature}: {price_correlation[feature]:.4f}")

# 对每个特征进行详细分析
for feature in top3_features:
    print(f"\n===== {feature} 分析 =====")
    
    # 查看特征描述（从数据说明文档中提取）
    descriptions = {
        'OverallQual': '房屋整体材质和装修质量，1-10分，10分为最好',
        'GrLivArea': '地上居住面积（平方英尺）',
        'GarageCars': '车库可容纳的车辆数量'
    }
    print(f"特征描述: {descriptions.get(feature, '无可用描述')}")
    
    # 分组统计
    if feature == 'OverallQual':  # 质量是离散值，可以直接分组
        grouped = df.groupby(feature)['SalePrice'].agg(['mean', 'count'])
    else:  # 对连续值进行分组
        # 对特征进行去重（根据实际情况选择是否保留重复值）
        unique_feature = df[feature].drop_duplicates()
# 然后再进行分箱
        grouped = df.groupby(pd.qcut(unique_feature, 5))['SalePrice'].agg(['mean', 'count'])
    
    print("\n按特征分组的房价均值:")
    display(grouped)
    
    # 可视化特征与房价的关系
    plt.figure(figsize=(10, 6))
    if feature == 'OverallQual':
        sns.boxplot(x=feature, y='SalePrice', data=df)
    else:
        sns.regplot(x=feature, y='SalePrice', data=df, scatter_kws={'alpha':0.5})
    plt.title(f'{feature} 与 房价 的关系')
    plt.tight_layout()
    plt.show()

# 综合解释
print("\n===== 相关性分析总结 =====")
print("1. OverallQual（房屋整体质量）与房价的相关性最高，这符合常理，因为房屋质量直接决定了其价值。")
print("2. GrLivArea（地上居住面积）是第二个重要因素，更大的居住空间通常意味着更高的价格。")
print("3. GarageCars（车库容量）作为第三个重要因素，反映了房屋的附加价值和便利性，对房价有显著影响。")
print("\n这些发现符合房地产市场的一般规律：质量、面积和便利设施是决定房屋价格的三大核心因素。")
## 9. 保存处理后的数据集# 保存处理后的数据集
df.to_csv('processed_housing_data.csv', index=False)
print("处理后的数据集已保存为 'processed_housing_data.csv'")
```
